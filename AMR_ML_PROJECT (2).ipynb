```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

categorical_columns = [
    "Species",
    "Country",
    "Gender",
    "Age Group",
    "Speciality",
    "Source"
]

# Fit encoder using only the 50k training data
preprocessor_50000 = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_columns
        )
    ],
    remainder="passthrough"
)

X_dev_50000_encoded = preprocessor_50000.fit_transform(X_dev_50000)
X_test_50000_encoded = preprocessor_50000.transform(X_test)

print("Encoded training shape:", X_dev_50000_encoded.shape)
print("Encoded test shape:", X_test_50000_encoded.shape)
```

```python
from imblearn.over_sampling import SMOTE

smote_50000 = SMOTE(random_state=42)

# Apply SMOTE only to the training data
X_dev_50000_balanced, y_dev_50000_balanced = smote_50000.fit_resample(
    X_dev_50000_encoded,
    y_dev_50000
)

print("Before SMOTE:", y_dev_50000.value_counts().to_dict())
print("After SMOTE :", pd.Series(y_dev_50000_balanced).value_counts().to_dict())
```

```python
# Find the best threshold for the 50k model
thresholds_50000 = np.arange(0.20, 0.41, 0.01)

threshold_results_50000 = []

for threshold in thresholds_50000:
    y_pred = (y_prob_50000 >= threshold).astype(int)

    threshold_results_50000.append({
        "Threshold": round(threshold, 2),
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred),
        "Recall": recall_score(y_test, y_pred),
        "F1": f1_score(y_test, y_pred)
    })

threshold_results_50000 = pd.DataFrame(threshold_results_50000)

best_threshold_50000 = threshold_results_50000.loc[
    threshold_results_50000["F1"].idxmax()
]

print("Best 50k threshold:")
print(best_threshold_50000)
```

```python
# Separate antibiotic results, AST interpretations, and mechanism fields
interpretation_columns = [
    col for col in amr.columns
    if col.endswith("_I")
]

mechanism_keywords = [
    "KPC", "NDM", "OXA", "VIM", "IMP",
    "CTX", "ESBL", "AmpC", "MRSA", "VRE"
]

mechanism_columns = [
    col for col in amr.columns
    if any(keyword.lower() in col.lower() for keyword in mechanism_keywords)
]

antibiotic_columns = [
    col for col in amr.columns
    if col not in interpretation_columns
    and col not in mechanism_columns
    and col not in [
        "Isolate Id", "Study", "Species", "Country",
        "State", "Gender", "Age Group", "Speciality",
        "Source", "Year"
    ]
]

print("AST interpretation columns:", len(interpretation_columns))
print("\nMechanism-related columns:", len(mechanism_columns))
print(mechanism_columns)

print("\nPotential antibiotic/MIC columns:", len(antibiotic_columns))
print(antibiotic_columns)
```

```python
# Inspect the final columns to identify mechanism fields correctly
for i, col in enumerate(amr.columns):
    print(i, ":", col)
```

```python
# Show antibiotic and corresponding AST interpretation columns
ast_pairs = []

for col in amr.columns:
    if col.endswith("_I"):
        antibiotic = col[:-2]
        if antibiotic in amr.columns:
            ast_pairs.append((antibiotic, col))

print("Antibiotic–AST pairs:", len(ast_pairs))

for antibiotic, interpretation in ast_pairs:
    print(f"{antibiotic}  →  {interpretation}")
```

```python
# Calculate availability of other AST results
other_ast_columns = [
    interpretation
    for antibiotic, interpretation in ast_pairs
    if antibiotic != "Amikacin"
]

ast_coverage = (
    amr_amikacin[other_ast_columns]
    .notna()
    .mean()
    .sort_values(ascending=False)
    * 100
)

print("Other AST features:", len(other_ast_columns))
print("\nAST availability (%):")
print(ast_coverage.round(2))
```

```python
# Visualize AST availability
plt.figure(figsize=(10, 12))

ast_coverage.sort_values().plot(kind="barh")

plt.xlabel("Available observations (%)")
plt.ylabel("Antibiotic AST")
plt.title("Availability of Other Antibiotic AST Results")
plt.xlim(0, 100)
plt.tight_layout()
plt.show()
```

```python
# Keep other AST features with at least 50% availability
selected_ast_columns = ast_coverage[
    ast_coverage >= 50
].index.tolist()

print("Selected AST features:", len(selected_ast_columns))
print("\nSelected columns:")
print(selected_ast_columns)
```

```python
# Combine epidemiological features with other AST results
base_features = [
    "Species",
    "Country",
    "Gender",
    "Age Group",
    "Speciality",
    "Source",
    "Year"
]

enhanced_features = base_features + selected_ast_columns

X_enhanced = amr_amikacin[enhanced_features].copy()

print("Enhanced feature count:", X_enhanced.shape[1])
print("Enhanced dataset shape:", X_enhanced.shape)
```

```python
# Preserve missing AST results as a separate category
for col in selected_ast_columns:
    X_enhanced[col] = X_enhanced[col].fillna("Not Tested")

print("Missing AST values remaining:",
      X_enhanced[selected_ast_columns].isna().sum().sum())
```

```python
# Use the same 50k sampling strategy
X_enhanced_50000, _, y_enhanced_50000, _ = train_test_split(
    X_enhanced,
    y,
    train_size=50000,
    random_state=42,
    stratify=y
)

print("Enhanced 50k shape:", X_enhanced_50000.shape)
print("Target shape:", y_enhanced_50000.shape)
```

```python
# Use the same original train/test indices
X_enhanced_test = X_enhanced.loc[X_test.index].copy()

print("Enhanced training:", X_enhanced_50000.shape)
print("Enhanced test:", X_enhanced_test.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

enhanced_categorical_columns = [
    col for col in enhanced_features
    if col != "Year"
]

# Encode categorical and AST features
preprocessor_enhanced = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(handle_unknown="ignore"),
            enhanced_categorical_columns
        )
    ],
    remainder="passthrough"
)

X_enhanced_50000_encoded = preprocessor_enhanced.fit_transform(
    X_enhanced_50000
)

X_enhanced_test_encoded = preprocessor_enhanced.transform(
    X_enhanced_test
)

print(
    "Enhanced encoded training:",
    X_enhanced_50000_encoded.shape
)

print(
    "Enhanced encoded test:",
    X_enhanced_test_encoded.shape
)
```

```python
from imblearn.over_sampling import SMOTE

smote_enhanced = SMOTE(random_state=42)

# Balance only the enhanced training data
X_enhanced_balanced, y_enhanced_balanced = smote_enhanced.fit_resample(
    X_enhanced_50000_encoded,
    y_enhanced_50000
)

print(
    "Before SMOTE:",
    y_enhanced_50000.value_counts().to_dict()
)

print(
    "After SMOTE:",
    pd.Series(y_enhanced_balanced).value_counts().to_dict()
)
```

## Train the enhanced Random Forest

```python
# Train Random Forest with enhanced AST features
rf_enhanced = RandomForestClassifier(
    n_estimators=200,
    max_depth=None,
    max_features="sqrt",
    min_samples_split=5,
    min_samples_leaf=1,
    random_state=42,
    n_jobs=-1
)

rf_enhanced.fit(X_enhanced_balanced, y_enhanced_balanced)

y_prob_enhanced = rf_enhanced.predict_proba(
    X_enhanced_test_encoded
)[:, 1]
```

```python
# Evaluate enhanced model
y_pred_enhanced = (y_prob_enhanced >= 0.50).astype(int)

print(f"Accuracy : {accuracy_score(y_test, y_pred_enhanced):.4f}")
print(f"Precision: {precision_score(y_test, y_pred_enhanced):.4f}")
print(f"Recall   : {recall_score(y_test, y_pred_enhanced):.4f}")
print(f"F1 Score : {f1_score(y_test, y_pred_enhanced):.4f}")
print(f"ROC-AUC  : {roc_auc_score(y_test, y_prob_enhanced):.4f}")
```

```python
# Optimize threshold for the enhanced model
thresholds = np.arange(0.20, 0.61, 0.01)
results = []

for t in thresholds:
    pred = (y_prob_enhanced >= t).astype(int)
    results.append([t, precision_score(y_test, pred),
                    recall_score(y_test, pred), f1_score(y_test, pred)])

enhanced_thresholds = pd.DataFrame(
    results, columns=["Threshold", "Precision", "Recall", "F1"]
)

best_enhanced = enhanced_thresholds.loc[
    enhanced_thresholds["F1"].idxmax()
]

print(best_enhanced)
```

```python
# Reload the AMR dataset and recreate the Amikacin analysis subset
import numpy as np
import pandas as pd

amr = pd.read_csv("atlas_vivli_2004_2024.csv", low_memory=False)

amr = amr.replace(r"^\s*$", np.nan, regex=True)

amr_amikacin = amr[
    amr["Amikacin_I"].isin(["Susceptible", "Resistant"])
].copy()

y = amr_amikacin["Amikacin_I"].map({
    "Susceptible": 0,
    "Resistant": 1
})

print(amr_amikacin.shape)
```

```python
# Recreate the enhanced feature set using the previously selected AST features
feature_columns = [
    "Species", "Country", "Gender",
    "Age Group", "Speciality", "Source", "Year"
]

ast_pairs = []

for col in amr.columns:
    if col.endswith("_I"):
        antibiotic = col[:-2]
        if antibiotic in amr.columns:
            ast_pairs.append((antibiotic, col))

ast_coverage = amr_amikacin[
    [ast for _, ast in ast_pairs if ast != "Amikacin_I"]
].notna().mean() * 100

selected_ast_columns = ast_coverage[
    ast_coverage >= 50
].index.tolist()

enhanced_features = feature_columns + selected_ast_columns

X_enhanced = amr_amikacin[enhanced_features].copy()

for col in selected_ast_columns:
    X_enhanced[col] = X_enhanced[col].fillna("Not Tested")

print("Enhanced features:", len(enhanced_features))
print(X_enhanced.shape)
```

```python
# Recreate the 70/15/15 train-validation-final-test split
from sklearn.model_selection import train_test_split

X_train_full, X_final_test, y_train_full, y_final_test = train_test_split(
    X_enhanced, y,
    test_size=0.15,
    random_state=42,
    stratify=y
)

X_train, X_val, y_train, y_val = train_test_split(
    X_train_full, y_train_full,
    test_size=0.1765,
    random_state=42,
    stratify=y_train_full
)

print("Train:", X_train.shape)
print("Validation:", X_val.shape)
print("Final test:", X_final_test.shape)
```

```python
# Fit the encoder on training data only and transform validation/test data
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

enhanced_categorical_columns = [
    col for col in X_train.columns if col != "Year"
]

preprocessor_final = ColumnTransformer(
    transformers=[
        ("categorical",
         OneHotEncoder(handle_unknown="ignore"),
         enhanced_categorical_columns)
    ],
    remainder="passthrough"
)

X_train_encoded = preprocessor_final.fit_transform(X_train)
X_val_encoded = preprocessor_final.transform(X_val)
X_final_test_encoded = preprocessor_final.transform(X_final_test)

print("Train encoded:", X_train_encoded.shape)
print("Validation encoded:", X_val_encoded.shape)
print("Final test encoded:", X_final_test_encoded.shape)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

# Recreate RF predictions on the final test set
final_threshold = 0.44

y_prob_rf_final = rf_final.predict_proba(
    X_final_test_encoded
)[:, 1]

y_pred_rf_final = (
    y_prob_rf_final >= final_threshold
).astype(int)

# Create evaluation dataset
species_test = X_final_test.copy()

species_test["Actual"] = y_final_test.values
species_test["Predicted"] = y_pred_rf_final
species_test["Probability"] = y_prob_rf_final

print("Species evaluation dataset recreated.")
print("Shape:", species_test.shape)
```

```python
# Calculate species-wise model performance

species_results = []

species_counts = species_test["Species"].value_counts()

eligible_species = species_counts[
    species_counts >= 500
].index

for species in eligible_species:

    data = species_test[
        species_test["Species"] == species
    ]

    if data["Actual"].nunique() < 2:
        continue

    species_results.append({
        "Species": species,
        "Samples": len(data),
        "Accuracy": accuracy_score(
            data["Actual"],
            data["Predicted"]
        ),
        "Precision": precision_score(
            data["Actual"],
            data["Predicted"],
            zero_division=0
        ),
        "Recall": recall_score(
            data["Actual"],
            data["Predicted"],
            zero_division=0
        ),
        "F1": f1_score(
            data["Actual"],
            data["Predicted"],
            zero_division=0
        ),
        "ROC-AUC": roc_auc_score(
            data["Actual"],
            data["Probability"]
        )
    })

species_results_df = pd.DataFrame(
    species_results
)

species_results_df = species_results_df.sort_values(
    "Samples",
    ascending=False
)

display(species_results_df.round(4))
```

```python
# Calculate resistance rate for each bacterial species

species_resistance = (
    species_test
    .groupby("Species")["Actual"]
    .agg(
        Samples="count",
        Resistant="sum"
    )
    .reset_index()
)

species_resistance["Resistance_Rate"] = (
    species_resistance["Resistant"]
    / species_resistance["Samples"]
    * 100
)

species_resistance = species_resistance[
    species_resistance["Samples"] >= 500
]

species_resistance = species_resistance.sort_values(
    "Resistance_Rate",
    ascending=False
)

display(species_resistance.round(2))
```

```python
# Compare available ESKAPE-related species

eskape_species = [
    "Acinetobacter baumannii",
    "Klebsiella pneumoniae",
    "Pseudomonas aeruginosa",
    "Enterobacter cloacae",
    "Enterobacter hormaechi"
]

eskape_results = []

for species in eskape_species:

    data = species_test[
        species_test["Species"] == species
    ]

    if len(data) == 0:
        continue

    if data["Actual"].nunique() < 2:
        continue

    eskape_results.append({
        "Species": species,
        "Samples": len(data),
        "Resistance Rate (%)": data["Actual"].mean() * 100,
        "Accuracy": accuracy_score(
            data["Actual"],
            data["Predicted"]
        ),
        "Precision": precision_score(
            data["Actual"],
            data["Predicted"],
            zero_division=0
        ),
        "Recall": recall_score(
            data["Actual"],
            data["Predicted"],
            zero_division=0
        ),
        "F1": f1_score(
            data["Actual"],
            data["Predicted"],
            zero_division=0
        ),
        "ROC-AUC": roc_auc_score(
            data["Actual"],
            data["Probability"]
        )
    })

eskape_df = pd.DataFrame(eskape_results)

eskape_df = eskape_df.sort_values(
    "Resistance Rate (%)",
    ascending=False
)

display(eskape_df.round(2))
```

```python
import streamlit as st
import pandas as pd
import joblib


# Load the trained model, preprocessing pipeline, and decision threshold
@st.cache_resource
def load_model():

    model = joblib.load("rf_final.pkl")
    preprocessor = joblib.load("preprocessor_final.pkl")
    threshold = joblib.load("rf_threshold.pkl")

    return model, preprocessor, threshold


rf_model, preprocessor, threshold = load_model()


# Page configuration
st.set_page_config(
    page_title="AMR Prediction System",
    page_icon="🧬",
    layout="wide"
)


st.title("AMR Prediction System")

st.write(
    "Machine learning prototype for predicting "
    "Amikacin resistance."
)

st.warning(
    "This is an academic decision-support prototype "
    "and not a clinical diagnostic system."
)


st.header("Sample Information")

col1, col2 = st.columns(2)


with col1:

    species = st.selectbox(
        "Species",
        [
            "Acinetobacter baumannii",
            "Klebsiella pneumoniae",
            "Proteus mirabilis",
            "Pseudomonas aeruginosa",
            "Serratia marcescens",
            "Escherichia coli",
            "Enterobacter hormaechi",
            "Enterobacter cloacae",
            "Citrobacter freundii",
            "Morganella morganii",
            "Klebsiella aerogenes",
            "Acinetobacter pitii",
            "Proteus vulgaris",
            "Klebsiella oxytoca",
            "Citrobacter koseri"
        ]
    )

    country = st.text_input(
        "Country",
        value="India"
    )

    gender = st.selectbox(
        "Gender",
        [
            "Male",
            "Female",
            "Unknown"
        ]
    )

    age_group = st.text_input(
        "Age Group",
        value="Adult"
    )


with col2:

    speciality = st.text_input(
        "Speciality",
        value="General"
    )

    source = st.text_input(
        "Source",
        value="Unknown"
    )

    year = st.number_input(
        "Year",
        min_value=2000,
        max_value=2030,
        value=2025,
        step=1
    )


st.header("Antibiotic Susceptibility")

st.write(
    "Enter the observed R/S result for each antibiotic."
)


antibiotics = [
    "Amoxycillin clavulanate_I",
    "Ampicillin_I",
    "Aztreonam_I",
    "Cefepime_I",
    "Ceftazidime avibactam_I",
    "Ceftazidime_I",
    "Colistin_I",
    "Imipenem_I",
    "Levofloxacin_I",
    "Meropenem_I",
    "Piperacillin tazobactam_I",
    "Tigecycline_I"
]


antibiotic_values = {}

columns = st.columns(3)


for index, antibiotic in enumerate(antibiotics):

    with columns[index % 3]:

        antibiotic_values[antibiotic] = st.selectbox(
            antibiotic.replace("_I", ""),
            ["S", "R"],
            key=antibiotic
        )


st.divider()


if st.button(
    "Predict Amikacin Resistance",
    type="primary"
):

    input_data = {
        "Species": species,
        "Country": country,
        "Gender": gender,
        "Age Group": age_group,
        "Speciality": speciality,
        "Source": source,
        "Year": year
    }

    input_data.update(antibiotic_values)

    input_df = pd.DataFrame([input_data])

    try:

        encoded_input = preprocessor.transform(
            input_df
        )

        probability = rf_model.predict_proba(
            encoded_input
        )[0, 1]

        prediction = int(
            probability >= threshold
        )

        st.header("Prediction Result")

        if prediction == 1:

            st.error(
                "Predicted Result: Resistant"
            )

        else:

            st.success(
                "Predicted Result: Susceptible"
            )

        st.metric(
            "Resistance Probability",
            f"{probability * 100:.2f}%"
        )

        st.caption(
            f"Decision threshold: {threshold:.2f}"
        )

    except Exception as error:

        st.error(
            f"Prediction error: {error}"
        )
```
